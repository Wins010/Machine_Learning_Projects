### d. Computing the slope 'm' and intercept 'b'

### Simple Linear Regression Formula

The regression line is:

$$
y = m x + b
$$

where

$$
m = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2}
\qquad
b = \bar{y} - m\,\bar{x}
$$

- $m$ = slope (coefficient)  
- $b$ = intercept
